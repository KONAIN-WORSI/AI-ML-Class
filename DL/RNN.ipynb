```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
```

```python
sentences = [
    # Positive sentences
    "I am feeling happy today.",
    "The weather is beautiful.",
    "I enjoy spending time with friends.",
    "She is very kind.",
    "He always helps others.",
    "I had a great day.",
    "The food tastes delicious.",
    "We are excited about the trip.",
    "My new laptop works perfectly.",
    "I love learning new things.",
    "The movie was really good.",
    "She looks happy today.",
    "I am proud of my work.",
    "Everything is going well.",
    "I feel confident about the future.",

    # Negative sentences
    "I am feeling tired today.",
    "The weather is terrible.",
    "I don't like this food.",
    "She is not feeling well.",
    "He never helps anyone.",
    "I had a bad day.",
    "The food tastes awful.",
    "We are worried about the trip.",
    "My laptop is not working.",
    "I don't understand this topic.",
    "The movie was boring.",
    "She looks unhappy today.",
    "I am not satisfied with my work.",
    "Everything is going wrong.",
    "I feel worried about the future."
]

labels = [1]*15 + [0]*15
labels = np.array(labels)
```

```python
vocab_size = 2000
tok = Tokenizer(num_words = vocab_size, oov_token = '<OOV>')
tok.fit_on_texts(sentences)
seqs = tok.texts_to_sequences(sentences)
max_len = max(len(s) for s in seqs)
X = pad_sequences(seqs, maxlen = max_len, padding = 'post')
y = labels 
```

```python
X[1]
```

```python
embeded_dim = 16
rnn_unit = 8
```

```python
inp = Input(shape = (max_len, ), dtype='int32', name='input')
x = Embedding(input_dim=vocab_size, output_dim=embeded_dim, mask_zero=True, name='embed')(inp)
rnn = SimpleRNN(units = rnn_unit, return_sequences = False, return_state=False, name= 'simple_rnn')
x_last = rnn(x)
out = Dense(1, activation = 'sigmoid', name= 'out')(x_last)
model = Model(inputs = inp, outputs = out)
model.compile(optimizer='adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
model.summary()
```

```python
model.fit(X, y, epochs = 25, batch_size = 8, verbose = 1)
```
