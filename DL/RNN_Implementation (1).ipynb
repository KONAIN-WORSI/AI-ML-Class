```python
import string

translator = str.maketrans('', '', string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size = 10000

tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index = tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequences = tokenizer.texts_to_sequences(quotes)
print(sequences[:5])
```

```python
X = []
y = []

for seq in sequences:
    for i in range(1, len(seq)):
        input_shape = seq[:i]
        output_seq = seq[i]
        X.append(input_shape)
        y.append(output_seq)
```

```python
max_len = max(len(x) for x in X)
print(max_len)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded = pad_sequences(X, maxlen=max_len, padding='pre')
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y, num_classes = vocab_size)
```

```python
y_one_hot.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, SimpleRNN, GRU
```

```python
embeddings_dim = 50
rnn_units = 128
```

```python
rnn_model = Sequential()
rnn_model.add(Embedding(input_dim=vocab_size, output_dim=embeddings_dim, input_length=max_len))
rnn_model.add(SimpleRNN(units=rnn_units))
rnn_model.add(Dense(units=vocab_size, activation='softmax'))
```

```python
rnn_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
rnn_model.summary()
```

```python
lstm_model = Sequential()
lstm_model.add(Embedding(input_dim=vocab_size, output_dim=embeddings_dim, input_length=max_len))
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=vocab_size, activation='softmax'))
```

```python
lstm_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
lstm_model.summary()
```

```python
gru_model = Sequential()
gru_model.add(Embedding(input_dim=vocab_size, output_dim=embeddings_dim, input_length=max_len))
gru_model.add(GRU(units=rnn_units))
gru_model.add(Dense(units=vocab_size, activation='softmax'))
```

```python
gru_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
gru_model.summary()
```

```python
index_to_word = {}

for word, index in word_index.items():
  index_to_word[index] = word
```

```python
index_to_word
```

```python
def predictor(model, tokenizer, text, max_len):
  text = text.lower()
  # Wrap text in a list so it tokenizes words instead of characters
  seq_list = tokenizer.texts_to_sequences([text])[0]
  # Pad the extracted 1D list of tokens
  padded_seq = pad_sequences([seq_list], maxlen=max_len, padding='pre')

  predict = model.predict(padded_seq, verbose=0)
  predict_index = np.argmax(predict)
  predicted_word = index_to_word[predict_index]

  return predicted_word
```

```python
seed_word = "World is"
next_word = predictor(gru_model, tokenizer, seed_word, max_len)
print(next_word)
```

```python
def generate_text(model, tokenizer, seed_text, max_len, n_words):
  for word in range(n_words):
    next_word = predictor(model, tokenizer, seed_text, max_len)
    if next_word == "":
      break
    seed_text += " " + next_word

  return seed_text
```

```python
seed = "God is"
# Use a different variable name to avoid overwriting the function definition
generated_output = generate_text(gru_model, tokenizer, seed, max_len, 10)
print(generated_output)
```
